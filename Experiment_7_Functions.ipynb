## Part 3: Functions with Return Values

```python
# Program 3: Functions with Return Values
print("\n" + "=" * 70)
print("Program 3: Functions with Return Values")
print("=" * 70)

# 1. Function that returns a single value
def square(number):
    """Function that returns the square of a number."""
    result = number ** 2
    return result

print("\n1. Function returning single value:")
print("-" * 70)
result1 = square(5)
print(f"square(5) = {result1}")
result2 = square(10)
print(f"square(10) = {result2}")
result3 = square(-3)
print(f"square(-3) = {result3}")

# 2. Function that returns multiple values
def calculate_operations(a, b):
    """Function that returns multiple calculation results."""
    addition = a + b
    subtraction = a - b
    multiplication = a * b
    if b != 0:
        division = a / b
    else:
        division = "Cannot divide by zero"
    return addition, subtraction, multiplication, division

print("\n2. Function returning multiple values:")
print("-" * 70)
add, sub, mul, div = calculate_operations(10, 5)
print(f"calculate_operations(10, 5):")
print(f"  Addition: {add}")
print(f"  Subtraction: {sub}")
print(f"  Multiplication: {mul}")
print(f"  Division: {div}")

print("Another example:")
add, sub, mul, div = calculate_operations(20, 0)
print(f"calculate_operations(20, 0):")
print(f"  Addition: {add}")
print(f"  Subtraction: {sub}")
print(f"  Multiplication: {mul}")
print(f"  Division: {div}")

# 3. Function that returns different types
def check_number_type(num):
    """Function that returns different information based on number type."""
    if num > 0:
        return "Positive", num, num % 2 == 0
    elif num < 0:
        return "Negative", num, num % 2 == 0
    else:
        return "Zero", num, True  # Zero is even

print("\n3. Function returning different types:")
print("-" * 70)
for test_num in [5, -3, 0, 12, -8]:
    result_type, number, is_even = check_number_type(test_num)
    even_odd = "even" if is_even else "odd"
    print(f"{test_num}: {result_type}, {even_odd}")

# 4. Function that returns None (implicit return)
def display_message(message):
    """Function that displays a message but doesn't return anything."""
    print(f"Message: {message}")
    # No return statement - returns None implicitly

print("\n4. Function with no return statement:")
print("-" * 70)
result = display_message("Hello from function!")
print(f"Return value: {result}")
print(f"Type of return value: {type(result)}")

# 5. Early return
def find_first_even(numbers):
    """Function that returns the first even number or None if not found."""
    for num in numbers:
        if num % 2 == 0:
            return num  # Early return
    return None  # Return None if no even number found

print("\n5. Early return in function:")
print("-" * 70)
test_lists = [[1, 3, 5, 7], [1, 2, 3, 4], [1, 3, 5, 8, 9], [1, 3, 5]]
for lst in test_lists:
    first_even = find_first_even(lst)
    print(f"List: {lst} -> First even: {first_even}")

# 6. Using return values in expressions
def get_grade(marks):
    """Function that returns grade based on marks."""
    if marks >= 90:
        return 'A'
    elif marks >= 80:
        return 'B'
    elif marks >= 70:
        return 'C'
    elif marks >= 60:
        return 'D'
    else:
        return 'F'

print("\n6. Using return values in expressions:")
print("-" * 70)
student_marks = [95, 85, 75, 65, 55]
for marks in student_marks:
    grade = get_grade(marks)
    status = "Pass" if grade != 'F' else "Fail"
    print(f"Marks: {marks} -> Grade: {grade} -> Status: {status}")
```

## Part 4: Default Arguments

```python
# Program 4: Default Arguments
print("\n" + "=" * 70)
print("Program 4: Default Arguments")
print("=" * 70)

# 1. Function with default arguments
def greet_with_default(name, greeting="Hello", punctuation="!"):
    """Function with default arguments."""
    print(f"{greeting}, {name}{punctuation}")

print("\n1. Function with default arguments:")
print("-" * 70)
# Using all defaults
greet_with_default("Abhay")
# Using custom greeting
greet_with_default("Alice", "Hi")
# Using custom greeting and punctuation
greet_with_default("Bob", "Hey", "?")
# Using positional and keyword arguments
greet_with_default("Charlie", punctuation="!!!")

# 2. Calculator with default operation
def calculate(a, b, operation="add"):
    """Calculator function with default operation."""
    if operation == "add":
        result = a + b
        symbol = "+"
    elif operation == "subtract":
        result = a - b
        symbol = "-"
    elif operation == "multiply":
        result = a * b
        symbol = "*"
    elif operation == "divide":
        if b != 0:
            result = a / b
            symbol = "/"
        else:
            return "Error: Division by zero"
    else:
        return "Error: Invalid operation"
    
    return f"{a} {symbol} {b} = {result}"

print("\n2. Calculator with default operation:")
print("-" * 70)
print(f"calculate(10, 5): {calculate(10, 5)}")
print(f"calculate(10, 5, 'subtract'): {calculate(10, 5, 'subtract')}")
print(f"calculate(10, 5, 'multiply'): {calculate(10, 5, 'multiply')}")
print(f"calculate(10, 5, 'divide'): {calculate(10, 5, 'divide')}")
print(f"calculate(10, 0, 'divide'): {calculate(10, 0, 'divide')}")

# 3. Student info with defaults
def create_student_record(name, sap_id, course="Python", semester=1, grade="Not assigned"):
    """Create student record with default values."""
    return {
        'name': name,
        'sap_id': sap_id,
        'course': course,
        'semester': semester,
        'grade': grade
    }

print("\n3. Student record with defaults:")
print("-" * 70)
# Using all defaults
student1 = create_student_record("Abhay Rana", 590024916)
print(f"Student 1: {student1}")
# Custom course and semester
student2 = create_student_record("Alice Smith", 590012345, "Java", 2)
print(f"Student 2: {student2}")
# All custom values
student3 = create_student_record("Bob Johnson", 590023456, "C++", 3, "A")
print(f"Student 3: {student3}")

# 4. Important: Default arguments are evaluated once
def add_to_list(item, my_list=[]):
    """Demonstrates mutable default argument issue."""
    my_list.append(item)
    return my_list

print("\n4. Mutable default arguments (CAUTION):")
print("-" * 70)
print("This demonstrates why mutable defaults can be problematic:")
print(f"add_to_list('a'): {add_to_list('a')}")
print(f"add_to_list('b'): {add_to_list('b')}")
print(f"add_to_list('c'): {add_to_list('c')}")
print("Notice how the list persists between function calls!")

# Better approach
def add_to_list_safe(item, my_list=None):
    """Safe way to handle mutable default arguments."""
    if my_list is None:
        my_list = []
    my_list.append(item)
    return my_list

print("\nBetter approach with None default:")
print(f"add_to_list_safe('x'): {add_to_list_safe('x')}")
print(f"add_to_list_safe('y'): {add_to_list_safe('y')}")
print(f"add_to_list_safe('z'): {add_to_list_safe('z')}")
print("Each call creates a new list.")

# 5. Default arguments with *args and **kwargs
def flexible_function(required_arg, *args, default_value="default", **kwargs):
    """Function demonstrating all parameter types together."""
    print(f"Required: {required_arg}")
    print(f"Args: {args}")
    print(f"Default: {default_value}")
    print(f"Kwargs: {kwargs}")

print("\n5. All parameter types together:")
print("-" * 70)
flexible_function("required")
print()
flexible_function("required", 1, 2, 3, default_value="custom", key1="value1", key2="value2")
```

## Part 5: Practical Functions

```python
# Program 5: Practical Functions
print("\n" + "=" * 70)
print("Program 5: Practical Functions")
print("=" * 70)

# 1. Addition function
def add_numbers(a, b):
    """Function to add two numbers."""
    return a + b

print("\n1. Addition Function:")
print("-" * 70)
print(f"add_numbers(5, 3) = {add_numbers(5, 3)}")
print(f"add_numbers(10.5, 2.3) = {add_numbers(10.5, 2.3)}")
print(f"add_numbers(-5, 8) = {add_numbers(-5, 8)}")

# 2. Factorial function
def factorial(n):
    """Function to calculate factorial of a number."""
    if n < 0:
        return "Error: Factorial not defined for negative numbers"
    elif n == 0 or n == 1:
        return 1
    else:
        result = 1
        for i in range(2, n + 1):
            result *= i
        return result

print("\n2. Factorial Function:")
print("-" * 70)
for num in [0, 1, 5, 7, 10, -2]:
    print(f"factorial({num}) = {factorial(num)}")

# 3. Prime number checker
def is_prime(n):
    """Function to check if a number is prime."""
    if n <= 1:
        return False
    if n <= 3:
        return True
    if n % 2 == 0 or n % 3 == 0:
        return False
    
    i = 5
    while i * i <= n:
        if n % i == 0 or n % (i + 2) == 0:
            return False
        i += 6
    return True

print("\n3. Prime Number Checker:")
print("-" * 70)
test_numbers = [2, 3, 4, 5, 17, 18, 23, 24, 29, 1, 0, -5]
for num in test_numbers:
    result = is_prime(num)
    status = "is prime" if result else "is not prime"
    print(f"{num} {status}")

# 4. Find maximum of two numbers
def find_max(a, b):
    """Function to find maximum of two numbers."""
    return a if a > b else b

print("\n4. Find Maximum of Two Numbers:")
print("-" * 70)
pairs = [(5, 3), (10, 20), (7.5, 7.5), (-5, -3), (0, 0)]
for a, b in pairs:
    max_val = find_max(a, b)
    print(f"max({a}, {b}) = {max_val}")

# 5. Find maximum of three numbers
def find_max_three(a, b, c):
    """Function to find maximum of three numbers."""
    return max(a, b, c)

print("\n5. Find Maximum of Three Numbers:")
print("-" * 70)
triplets = [(5, 3, 8), (10, 20, 15), (7.5, 7.5, 7.5), (-5, -3, -8)]
for a, b, c in triplets:
    max_val = find_max_three(a, b, c)
    print(f"max({a}, {b}, {c}) = {max_val}")

# 6. Temperature converter
def celsius_to_fahrenheit(celsius):
    """Convert Celsius to Fahrenheit."""
    return (celsius * 9/5) + 32

def fahrenheit_to_celsius(fahrenheit):
    """Convert Fahrenheit to Celsius."""
    return (fahrenheit - 32) * 5/9

print("\n6. Temperature Converter:")
print("-" * 70)
temps_c = [0, 25, 37, 100]
print("Celsius to Fahrenheit:")
for c in temps_c:
    f = celsius_to_fahrenheit(c)
    print(f"{c}°C = {f:.1f}°F")

print("\nFahrenheit to Celsius:")
temps_f = [32, 77, 98.6, 212]
for f in temps_f:
    c = fahrenheit_to_celsius(f)
    print(f"{f}°F = {c:.1f}°C")

# 7. String utilities
def count_vowels(text):
    """Count vowels in a string."""
    vowels = "aeiouAEIOU"
    return sum(1 for char in text if char in vowels)

def reverse_string(text):
    """Reverse a string."""
    return text[::-1]

def is_palindrome(text):
    """Check if string is palindrome."""
    text = text.lower().replace(" ", "")
    return text == text[::-1]

print("\n7. String Utilities:")
print("-" * 70)
test_strings = ["Hello World", "Python Programming", "racecar", "Madam", "12321"]
for s in test_strings:
    vowels = count_vowels(s)
    reversed_s = reverse_string(s)
    palindrome = is_palindrome(s)
    print(f"String: '{s}'")
    print(f"  Vowels: {vowels}")
    print(f"  Reversed: '{reversed_s}'")
    print(f"  Palindrome: {palindrome}")
    print()
```

## Part 6: Local vs Global Variables

```python
# Program 6: Local vs Global Variables
print("\n" + "=" * 70)
print("Program 6: Local vs Global Variables")
print("=" * 70)

# Global variables
global_counter = 0
global_message = "Hello from global scope"

print("\nGLOBAL VARIABLES:")
print("-" * 70)
print(f"global_counter = {global_counter}")
print(f"global_message = '{global_message}'")

# 1. Function accessing global variables (read-only)
def read_global():
    """Function that reads global variables."""
    print(f"Inside read_global(): global_counter = {global_counter}")
    print(f"Inside read_global(): global_message = '{global_message}'")
    # Local variable with same name
    local_message = "Hello from local scope"
    print(f"Inside read_global(): local_message = '{local_message}'")

print("\n1. Reading global variables:")
print("-" * 70)
read_global()
print(f"Outside function: global_message = '{global_message}'")
print("Note: local_message is not accessible outside the function")

# 2. Function modifying global variables
def modify_global():
    """Function that modifies global variables."""
    global global_counter  # Declare global to modify
    global_counter += 1
    print(f"Inside modify_global(): global_counter = {global_counter}")

print("\n2. Modifying global variables:")
print("-" * 70)
print(f"Before function call: global_counter = {global_counter}")
modify_global()
print(f"After function call: global_counter = {global_counter}")

# 3. Local variables take precedence
def local_vs_global():
    """Demonstrates local vs global variable precedence."""
    global_counter = 100  # This creates a local variable!
    print(f"Inside function: global_counter = {global_counter} (local)")
    print(f"Inside function: global global_counter would be accessed with 'global' keyword")

print("\n3. Local variables take precedence:")
print("-" * 70)
print(f"Before function call: global_counter = {global_counter}")
local_vs_global()
print(f"After function call: global_counter = {global_counter} (unchanged)")

# 4. Using global keyword properly
def proper_global_usage():
    """Proper way to modify global variables."""
    global global_counter, global_message
    global_counter = 50
    global_message = "Modified by function"
    print(f"Inside function: global_counter = {global_counter}")
    print(f"Inside function: global_message = '{global_message}'")

print("\n4. Proper global variable modification:")
print("-" * 70)
print(f"Before: global_counter = {global_counter}, global_message = '{global_message}'")
proper_global_usage()
print(f"After: global_counter = {global_counter}, global_message = '{global_message}'")

# 5. Scope demonstration with nested functions
def outer_function():
    """Outer function with nested function."""
    outer_var = "I'm in outer function"
    
    def inner_function():
        """Inner function that accesses outer scope."""
        nonlocal outer_var  # Access outer function's variable
        outer_var = "Modified by inner function"
        inner_var = "I'm in inner function"
        print(f"Inner function: outer_var = '{outer_var}'")
        print(f"Inner function: inner_var = '{inner_var}'")
    
    print(f"Outer function (before inner): outer_var = '{outer_var}'")
    inner_function()
    print(f"Outer function (after inner): outer_var = '{outer_var}'")

print("\n5. Nested functions and nonlocal keyword:")
print("-" * 70)
outer_function()

# 6. Variable scope rules summary
print("\n6. VARIABLE SCOPE RULES:")
print("-" * 70)
print("1. Local scope: Variables defined inside a function")
print("2. Enclosing scope: Variables in outer functions (nonlocal)")
print("3. Global scope: Variables defined at module level (global)")
print("4. Built-in scope: Python's built-in functions and constants")
print()
print("LEGB Rule: Local → Enclosing → Global → Built-in")
print()
print("To modify global variables inside functions, use 'global' keyword")
print("To modify enclosing function variables, use 'nonlocal' keyword")

# 7. Best practices
print("\n7. BEST PRACTICES:")
print("-" * 70)
print("✓ Use local variables whenever possible")
print("✓ Avoid modifying global variables inside functions")
print("✓ Pass values as parameters instead of using globals")
print("✓ Return values instead of modifying globals")
print("✓ Use global only when absolutely necessary")
print("✓ Be careful with mutable default arguments")
```

## Part 7: Function Examples and Reusability

```python
# Program 7: Function Examples and Reusability
print("\n" + "=" * 70)
print("Program 7: Function Examples and Reusability")
print("=" * 70)

# 1. Reusable calculator function
def calculator():
    """Interactive calculator demonstrating function reusability."""
    print("\n" + "CALCULATOR")
    print("-" * 30)
    
    while True:
        try:
            num1 = float(input("Enter first number: "))
            operation = input("Enter operation (+, -, *, /) or 'quit': ")
            
            if operation.lower() == 'quit':
                break
                
            num2 = float(input("Enter second number: "))
            
            if operation == '+':
                result = add_numbers(num1, num2)
            elif operation == '-':
                result = num1 - num2
            elif operation == '*':
                result = num1 * num2
            elif operation == '/':
                if num2 != 0:
                    result = num1 / num2
                else:
                    print("Error: Division by zero!")
                    continue
            else:
                print("Invalid operation!")
                continue
            
            print(f"Result: {num1} {operation} {num2} = {result}")
            
        except ValueError:
            print("Error: Please enter valid numbers!")
    
    print("Calculator closed.")

# Note: We won't call calculator() here as it requires user input
print("\n1. Interactive Calculator Function:")
print("-" * 70)
print("Function defined but not called here (requires user input)")
print("You can call calculator() to use the interactive calculator")

# 2. Data processing functions
def process_student_data(students):
    """Process a list of student dictionaries."""
    total_marks = 0
    highest_marks = 0
    lowest_marks = 100
    pass_count = 0
    
    for student in students:
        marks = student['marks']
        total_marks += marks
        highest_marks = max(highest_marks, marks)
        lowest_marks = min(lowest_marks, marks)
        if marks >= 50:
            pass_count += 1
    
    average = total_marks / len(students)
    pass_percentage = (pass_count / len(students)) * 100
    
    return {
        'total_students': len(students),
        'average_marks': round(average, 2),
        'highest_marks': highest_marks,
        'lowest_marks': lowest_marks,
        'pass_count': pass_count,
        'pass_percentage': round(pass_percentage, 2)
    }

print("\n2. Data Processing Function:")
print("-" * 70)

# Sample student data
student_data = [
    {'name': 'Alice', 'marks': 85},
    {'name': 'Bob', 'marks': 92},
    {'name': 'Charlie', 'marks': 78},
    {'name': 'David', 'marks': 45},
    {'name': 'Eve', 'marks': 88}
]

results = process_student_data(student_data)
print("Student Data Analysis:")
for key, value in results.items():
    print(f"  {key.replace('_', ' ').title()}: {value}")

# 3. Function composition
def get_student_grades(students):
    """Get grades for all students using the get_grade function."""
    return [(student['name'], get_grade(student['marks'])) for student in students]

print("\n3. Function Composition:")
print("-" * 70)
grades = get_student_grades(student_data)
print("Student Grades:")
for name, grade in grades:
    print(f"  {name}: {grade}")

# 4. Recursive function
def fibonacci(n):
    """Calculate nth Fibonacci number using recursion."""
    if n <= 1:
        return n
    else:
        return fibonacci(n-1) + fibonacci(n-2)

print("\n4. Recursive Function (Fibonacci):")
print("-" * 70)
print("Fibonacci sequence:")
for i in range(10):
    print(f"  F({i}) = {fibonacci(i)}")

# 5. Function as parameter
def apply_operation(numbers, operation_func):
    """Apply a function to each number in a list."""
    return [operation_func(num) for num in numbers]

print("\n5. Function as Parameter:")
print("-" * 70)
numbers = [1, 2, 3, 4, 5]
print(f"Original numbers: {numbers}")
print(f"Squares: {apply_operation(numbers, lambda x: x**2)}")
print(f"Cubes: {apply_operation(numbers, lambda x: x**3)}")
print(f"Factorials: {apply_operation(numbers, factorial)}")

# 6. Function returning function
def create_multiplier(factor):
    """Return a function that multiplies by a given factor."""
    def multiplier(number):
        return number * factor
    return multiplier

print("\n6. Function Returning Function:")
print("-" * 70)
double = create_multiplier(2)
triple = create_multiplier(3)
print(f"double(5) = {double(5)}")
print(f"triple(5) = {triple(5)}")
print(f"double(10) = {double(10)}")
print(f"triple(10) = {triple(10)}")

# 7. Code reusability demonstration
print("\n7. CODE REUSABILITY DEMONSTRATION:")
print("-" * 70)
print("Using the same functions with different data:")

# Test the same functions with different data
test_data1 = [10, 20, 30, 40, 50]
test_data2 = [1.5, 2.5, 3.5, 4.5, 5.5]

print(f"Data 1: {test_data1}")
print(f"Sum: {sum(test_data1)}")
print(f"Max: {max(test_data1)}")
print(f"Min: {min(test_data1)}")
print(f"Average: {sum(test_data1)/len(test_data1):.2f}")

print(f"Data 2: {test_data2}")
print(f"Sum: {sum(test_data2)}")
print(f"Max: {max(test_data2)}")
print(f"Min: {min(test_data2)}")
print(f"Average: {sum(test_data2)/len(test_data2):.2f}")

print("\nSame functions, different data - that's reusability!")
```

#### 1. **Function Definition:**
   - **Syntax:** `def function_name(parameters):`
   - Function body must be indented
   - Can include docstrings with `"""documentation"""`
   - Functions are objects in Python

#### 2. **Function Calling:**
   - Use function name followed by parentheses
   - Pass arguments in the correct order
   - Can use positional or keyword arguments
   - Functions can be called multiple times

#### 3. **Parameters and Arguments:**
   - **Parameters:** Variables in function definition
   - **Arguments:** Values passed to function
   - **Positional arguments:** Passed in order
   - **Keyword arguments:** Passed with parameter names
   - **Default arguments:** Have default values
   - **Variable arguments:** `*args` for multiple positional, `**kwargs` for keyword

#### 4. **Return Statement:**
   - `return` exits function and returns value(s)
   - Can return single value, multiple values (tuple), or None
   - Functions without return return None implicitly
   - Can have multiple return statements (early return)

#### 5. **Variable Scope:**
   - **Local scope:** Variables inside function
   - **Global scope:** Variables outside all functions
   - **LEGB rule:** Local → Enclosing → Global → Built-in
   - Use `global` to modify global variables
   - Use `nonlocal` to modify enclosing function variables

#### 6. **Function Types:**
   - **Built-in functions:** `print()`, `len()`, `max()`
   - **User-defined functions:** Created with `def`
   - **Lambda functions:** Anonymous functions
   - **Recursive functions:** Call themselves

#### 7. **Function Features:**
   - **Reusability:** Write once, use many times
   - **Modularity:** Break complex tasks into smaller functions
   - **Abstraction:** Hide implementation details
   - **Testing:** Test functions independently
   - **Documentation:** Use docstrings
