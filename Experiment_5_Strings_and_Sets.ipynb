## Part 4: Character and Word Counting

```python
# Program 4: Character and Word Counting
print("\n" + "=" * 70)
print("Program 4: Character and Word Counting")
print("=" * 70)

# User input
sentence = input("\nEnter a sentence: ")

# 1. Total characters
print("\n1. CHARACTER COUNTS:")
print("-" * 70)
print(f"Total characters (with spaces): {len(sentence)}")
characters_no_space = sentence.replace(" ", "")
print(f"Total characters (without spaces): {len(characters_no_space)}")

# 2. Word count
print("\n2. WORD COUNT:")
print("-" * 70)
words = sentence.split()
print(f"Total words: {len(words)}")
print(f"Words: {words}")

# 3. Vowel count
print("\n3. VOWEL COUNT:")
print("-" * 70)
vowels = "aeiouAEIOU"
vowel_count = 0
vowel_dict = {}
for vowel in vowels:
    count = sentence.count(vowel)
    if count > 0:
        vowel_dict[vowel] = count
        vowel_count += count
        print(f"'{vowel}': {count}")
print(f"\nTotal vowels: {vowel_count}")

# 4. Consonant count
print("\n4. CONSONANT COUNT:")
print("-" * 70)
consonant_count = 0
for char in sentence:
    if char.isalpha() and char not in vowels:
        consonant_count += 1
print(f"Total consonants: {consonant_count}")

# 5. Digit count
print("\n5. DIGIT COUNT:")
print("-" * 70)
digit_count = sum(1 for char in sentence if char.isdigit())
print(f"Total digits: {digit_count}")

# 6. Special characters count
print("\n6. SPECIAL CHARACTERS COUNT:")
print("-" * 70)
special_count = 0
special_chars = []
for char in sentence:
    if not char.isalnum() and char != ' ':
        special_count += 1
        if char not in special_chars:
            special_chars.append(char)
print(f"Total special characters: {special_count}")
print(f"Special characters found: {special_chars}")

# Summary
print("\n" + "=" * 70)
print("SUMMARY:")
print("=" * 70)
print(f"Original string: {sentence}")
print(f"Total length: {len(sentence)}")
print(f"Words: {len(words)}")
print(f"Vowels: {vowel_count}")
print(f"Consonants: {consonant_count}")
print(f"Digits: {digit_count}")
print(f"Special characters: {special_count}")
```

## Part 5: Set Creation and Properties

```python
# Program 5: Set Creation and Properties
print("\n" + "=" * 70)
print("Program 5: Set Creation and Properties")
print("=" * 70)

# 1. Creating sets
print("\n1. CREATING SETS:")
print("-" * 70)

# Empty set (not {})
set1 = set()
print(f"Empty set: {set1}")
print(f"Type: {type(set1).__name__}")

# Set with elements
set2 = {1, 2, 3, 4, 5}
print(f"\nSet with numbers: {set2}")

# Set with strings
set3 = {'apple', 'banana', 'cherry', 'date'}
print(f"Set with fruits: {set3}")

# Set from list
list_with_duplicates = [1, 2, 2, 3, 3, 3, 4, 4, 4, 4]
set4 = set(list_with_duplicates)
print(f"\nList with duplicates: {list_with_duplicates}")
print(f"Set from list (duplicates removed): {set4}")

# 2. Duplicate removal
print("\n2. AUTOMATIC DUPLICATE REMOVAL:")
print("-" * 70)
values = ['A', 'B', 'A', 'C', 'B', 'D', 'A']
print(f"Original list: {values}")
print(f"List length: {len(values)}")
set_values = set(values)
print(f"Set: {set_values}")
print(f"Set length: {len(set_values)}")
print(f"\nDuplicates removed: {len(values) - len(set_values)} duplicates")

# 3. Sets are unordered
print("\n3. SETS ARE UNORDERED:")
print("-" * 70)
set_demo = {5, 1, 3, 2, 4}
print(f"Created set: {{5, 1, 3, 2, 4}}")
print(f"Displayed as: {set_demo}")
print("Note: Order may vary each time")

# 4. Set operations - add and remove
print("\n4. ADD AND REMOVE ELEMENTS:")
print("-" * 70)
set_ops = {1, 2, 3}
print(f"Original set: {set_ops}")

set_ops.add(4)
print(f"After add(4): {set_ops}")

set_ops.add(2)  # Adding duplicate
print(f"After add(2) - duplicate: {set_ops}")

set_ops.remove(2)
print(f"After remove(2): {set_ops}")

set_ops.discard(5)  # Discard won't raise error if element not found
print(f"After discard(5) - not present: {set_ops}")
print("Note: discard() doesn't raise error if element not found")
```

## Part 6: Set Operations

```python
# Program 6: Set Operations
print("\n" + "=" * 70)
print("Program 6: Set Operations (Union, Intersection, Difference)")
print("=" * 70)

# Create two sets
setA = {1, 2, 3, 4, 5}
setB = {4, 5, 6, 7, 8}

print(f"\nSet A: {setA}")
print(f"Set B: {setB}")

# 1. Union - All elements from both sets
print("\n1. UNION (A | B or A.union(B)):")
print("-" * 70)
print("Description: All unique elements from both sets")
union_result = setA | setB
print(f"A | B = {union_result}")
union_method = setA.union(setB)
print(f"A.union(B) = {union_method}")
print(f"Elements: {sorted(union_result)}")

# 2. Intersection - Common elements
print("\n2. INTERSECTION (A & B or A.intersection(B)):")
print("-" * 70)
print("Description: Only elements present in both sets")
intersection_result = setA & setB
print(f"A & B = {intersection_result}")
intersection_method = setA.intersection(setB)
print(f"A.intersection(B) = {intersection_method}")
print(f"Elements: {sorted(intersection_result)}")

# 3. Difference - Elements in A but not in B
print("\n3. DIFFERENCE (A - B or A.difference(B)):")
print("-" * 70)
print("Description: Elements in A but NOT in B")
difference_result = setA - setB
print(f"A - B = {difference_result}")
difference_method = setA.difference(setB)
print(f"A.difference(B) = {difference_method}")
print(f"Elements: {sorted(difference_result)}")

# 4. Symmetric difference - Elements not in both
print("\n4. SYMMETRIC DIFFERENCE (A ^ B or A.symmetric_difference(B)):")
print("-" * 70)
print("Description: Elements in A or B but NOT in both")
sym_diff_result = setA ^ setB
print(f"A ^ B = {sym_diff_result}")
sym_diff_method = setA.symmetric_difference(setB)
print(f"A.symmetric_difference(B) = {sym_diff_method}")
print(f"Elements: {sorted(sym_diff_result)}")

# 5. Subset and Superset
print("\n5. SUBSET AND SUPERSET:")
print("-" * 70)
setC = {2, 3, 4}
setD = {1, 2, 3, 4, 5, 6}
print(f"Set C: {setC}")
print(f"Set D: {setD}")
print(f"C.issubset(D): {setC.issubset(setD)} - C is subset of D")
print(f"D.issuperset(C): {setD.issuperset(setC)} - D is superset of C")
print(f"C.isdisjoint(setB): {setC.isdisjoint(setB)} - C and B have no common elements")

# 6. Multiple set operations
print("\n6. WORKING WITH MULTIPLE SETS:")
print("-" * 70)
set1 = {1, 2, 3}
set2 = {3, 4, 5}
set3 = {5, 6, 7}
print(f"Set 1: {set1}")
print(f"Set 2: {set2}")
print(f"Set 3: {set3}")
print(f"\nUnion of all three: {set1 | set2 | set3}")
print(f"Intersection of all three: {set1 & set2 & set3}")
```

```python
# Program 8: Practical Examples
print("\n" + "=" * 70)
print("Program 8: Practical Applications")
print("=" * 70)

# Example 1: Remove duplicate words
print("\n1. REMOVE DUPLICATE WORDS FROM SENTENCE:")
print("-" * 70)
sentence = "python is great python is powerful python is awesome"
words_list = sentence.split()
words_set = set(words_list)
print(f"Original: {sentence}")
print(f"Words list: {words_list}")
print(f"Unique words: {words_set}")
print(f"Unique words (sorted): {sorted(words_set)}")

# Example 2: Find common characters
print("\n2. FIND COMMON CHARACTERS IN STRINGS:")
print("-" * 70)
str1 = "PYTHON"
str2 = "PROGRAMMING"
set1 = set(str1)
set2 = set(str2)
common = set1 & set2
print(f"String 1: {str1} -> Characters: {set1}")
print(f"String 2: {str2} -> Characters: {set2}")
print(f"Common characters: {sorted(common)}")

# Example 3: Find unique elements
print("\n3. FIND UNIQUE ELEMENTS IN TWO LISTS:")
print("-" * 70)
list1 = [1, 2, 3, 4, 5]
list2 = [4, 5, 6, 7, 8]
set1 = set(list1)
set2 = set(list2)
print(f"List 1: {list1}")
print(f"List 2: {list2}")
print(f"\nIntersection (common): {sorted(set1 & set2)}")
print(f"Union (all): {sorted(set1 | set2)}")
print(f"In List1 only: {sorted(set1 - set2)}")
print(f"In List2 only: {sorted(set2 - set1)}")

# Example 4: String palindrome check
print("\n4. PALINDROME CHECK:")
print("-" * 70)
def is_palindrome(s):
    s = s.replace(" ", "").lower()
    return s == s[::-1]

Test_strings = ["racecar", "hello", "A man a plan a canal Panama", "python"]
for test in test_strings:
    result = is_palindrome(test)
    print(f"'{test}': {result}")

# Example 5: Count frequency of characters
print("\n5. CHARACTER FREQUENCY:")
print("-" * 70)
text = "mississippi"
freq_dict = {}
for char in text:
    freq_dict[char] = freq_dict.get(char, 0) + 1
print(f"String: {text}")
print(f"Character frequency: {freq_dict}")
for char, count in sorted(freq_dict.items(), key=lambda x: x[1], reverse=True):
    print(f"  '{char}': {count}")
```

## Key Observations and Learning Summary

### STRINGS - Key Points:

#### 1. **String Creation:**
   - Single quotes: `'Hello'`
   - Double quotes: `"Hello"`
   - Triple quotes: `"""Multi-line"""`
   - f-strings: `f"Value: {var}"`
   - Strings are IMMUTABLE (cannot be changed in place)

#### 2. **String Indexing:**
   - Zero-based indexing: `s[0]` is first character
   - Negative indexing: `s[-1]` is last character
   - Out of range access raises IndexError

#### 3. **String Slicing:**
   - `s[start:end]` - from start to end-1
   - `s[start:]` - from start to end
   - `s[:end]` - from beginning to end-1
   - `s[::step]` - every nth character
   - `s[::-1]` - reverse the string

#### 4. **Important String Methods:**
   - **Case conversion:** `upper()`, `lower()`, `capitalize()`, `title()`
   - **Cleaning:** `strip()`, `lstrip()`, `rstrip()`
   - **Searching:** `find()`, `count()`, `startswith()`, `endswith()`
   - **Manipulation:** `replace()`, `split()`, `join()`
   - **Checking:** `isdigit()`, `isalpha()`, `isalnum()`, `isspace()`

#### 5. **String Operations:**
   - Concatenation: `s1 + s2`
   - Repetition: `s * 3`
   - Length: `len(s)`
   - Membership: `'a' in s`

---

### SETS - Key Points:

#### 1. **Set Creation:**
   - `set()` - empty set
   - `{1, 2, 3}` - set with elements
   - `set([1, 2, 2, 3])` - from list
   - Sets automatically REMOVE DUPLICATES
   - Sets are UNORDERED collections

#### 2. **Set Properties:**
   - No indexing (not ordered)
   - No slicing
   - Mutable (can add/remove elements)
   - Faster membership checking than lists (O(1) vs O(n))
   - Cannot contain unhashable types (lists, dicts)

#### 3. **Set Operations:**
   - **Union:** `A | B` or `A.union(B)` - all unique elements
   - **Intersection:** `A & B` or `A.intersection(B)` - common elements
   - **Difference:** `A - B` or `A.difference(B)` - in A but not B
   - **Symmetric Difference:** `A ^ B` - in A or B but not both

#### 4. **Set Methods:**
   - Adding: `add()`, `update()`
   - Removing: `remove()`, `discard()`, `pop()`
   - Checking: `issubset()`, `issuperset()`, `isdisjoint()`

#### 5. **Set Relationships:**
   - Subset: `A.issubset(B)` - all elements of A are in B
   - Superset: `A.issuperset(B)` - B's elements are in A
   - Disjoint: `A.isdisjoint(B)` - no common elements

---

### LISTS vs SETS Comparison:

| Feature | List | Set |
|---------|------|-----|
| **Ordering** | Ordered | Unordered |
| **Duplicates** | Allowed | Automatically removed |
| **Indexing** | Yes (list[0]) | No |
| **Slicing** | Yes | No |
| **Membership** | O(n) - slower | O(1) - faster |
| **Syntax** | [1, 2, 3] | {1, 2, 3} |
| **Modification** | append, insert, remove | add, remove, discard |
| **Use When** | Order/duplicates matter | Uniqueness matters |
