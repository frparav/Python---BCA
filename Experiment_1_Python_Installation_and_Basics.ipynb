## Step-by-Step Installation of Python

### Step 1: Open the official Python website
Go to [https://www.python.org/downloads/](https://www.python.org/downloads/).

### Step 2: Download Python
Click the latest stable **Python 3** download button for Windows.

### Step 3: Run the installer
Open the downloaded installer file.

### Step 4: Important option
Before clicking install, select the checkbox **Add Python to PATH**.

### Step 5: Install Python
Click **Install Now** and wait for the installation process to finish.

### Step 6: Verify installation
Open Command Prompt and type:
```bash
python --version
```
If Python is installed correctly, the version number will be displayed.

### Step 7: Install Jupyter Notebook
Open Command Prompt and type:
```bash
pip install notebook
```

### Step 8: Open Jupyter Notebook
Run the following command in Command Prompt:
```bash
jupyter notebook
```
This will open Jupyter Notebook in the web browser.

### Step 9: Create a new notebook
In Jupyter Notebook, click **New > Python 3** to create a new notebook.

### Step 10: Save the notebook
Save the notebook with the name **Experiment_1_Python_Installation_and_Basics.ipynb**.
