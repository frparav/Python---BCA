## Part 1: Understanding for Loop

```python
# Program 1: Simple for Loop
print("=" * 60)
print("Program 1: Simple for Loop")
print("=" * 60)

# Iterate through a range
print("\nPrinting numbers from 1 to 5:")
for i in range(1, 6):
    print(f"Number: {i}")

print("\nIterate through a list:")
fruits = ["Apple", "Banana", "Cherry", "Date"]
for fruit in fruits:
    print(f"Fruit: {fruit}")

print("\nIterate through a string:")
word = "PYTHON"
for char in word:
    print(f"Character: {char}")
```

## Part 2: Understanding while Loop

```python
# Program 2: Simple while Loop
print("\n" + "=" * 60)
print("Program 2: Simple while Loop")
print("=" * 60)

# Using while loop
print("\nPrinting numbers from 1 to 5 using while loop:")
i = 1
while i <= 5:
    print(f"Number: {i}")
    i = i + 1

# Another example
print("\nTaking input until user enters 'quit':")
while True:
    user_input = input("Enter something (type 'quit' to exit): ")
    if user_input.lower() == 'quit':
        print("Exiting...")
        break
    else:
        print(f"You entered: {user_input}")
```

## Part 3: Print Numbers in Sequence

```python
# Program 3: Print Numbers in Different Sequences
print("\n" + "=" * 60)
print("Program 3: Print Numbers in Sequences")
print("=" * 60)

# Ascending sequence
print("\nAscending sequence (1 to 10):")
for num in range(1, 11):
    print(num, end=' ')
print("\n")

# Descending sequence
print("Descending sequence (10 to 1):")
for num in range(10, 0, -1):
    print(num, end=' ')
print("\n")

# Even numbers
print("Even numbers (1 to 20):")
for num in range(2, 21, 2):
    print(num, end=' ')
print("\n")

# Odd numbers
print("Odd numbers (1 to 20):")
for num in range(1, 20, 2):
    print(num, end=' ')
print("\n")

# User defined range
start = int(input("\nEnter start number: "))
end = int(input("Enter end number: "))
step = int(input("Enter step size: "))

print(f"\nSequence from {start} to {end} with step {step}:")
for num in range(start, end + 1, step):
    print(num, end=' ')
print("\n")
```

## Part 4: Finding Sum of Numbers

```python
# Program 4: Sum of Numbers using Loops
print("\n" + "=" * 60)
print("Program 4: Sum of Numbers using Loops")
print("=" * 60)

# Sum of numbers from 1 to n using for loop
n = int(input("\nEnter the value of n: "))
sum_for = 0

print(f"\nCalculating sum from 1 to {n} using for loop:")
for i in range(1, n + 1):
    sum_for = sum_for + i
    print(f"Iteration {i}: sum = {sum_for}")

print(f"\nTotal sum (using for loop): {sum_for}")

# Sum of numbers using while loop
sum_while = 0
i = 1

print(f"\nCalculating sum from 1 to {n} using while loop:")
while i <= n:
    sum_while = sum_while + i
    print(f"Iteration {i}: sum = {sum_while}")
    i = i + 1

print(f"\nTotal sum (using while loop): {sum_while}")

# Sum of list of numbers
numbers = [10, 20, 30, 40, 50]
print(f"\nSum of list {numbers}:")
sum_list = 0
for num in numbers:
    sum_list += num
print(f"Sum: {sum_list}")

# Using Python's built-in sum function
print(f"\nUsing built-in sum function: {sum(numbers)}")
```

## Part 5: Multiplication Tables

```python
# Program 5: Multiplication Tables
print("\n" + "=" * 60)
print("Program 5: Multiplication Tables")
print("=" * 60)

# Single multiplication table
num = int(input("\nEnter a number for multiplication table: "))
print(f"\nMultiplication table of {num}:")
print("-" * 30)

for i in range(1, 11):
    product = num * i
    print(f"{num} × {i} = {product}")

# Multiple multiplication tables
print("\n" + "-" * 60)
print("Multiplication tables from 1 to 5:")
print("-" * 60)

for table_num in range(1, 6):
    print(f"\nMultiplication table of {table_num}:")
    for i in range(1, 6):
        print(f"{table_num} × {i} = {table_num * i}", end="   ")
    print()
```

## Part 6: Nested Loops for Patterns

```python
# Program 6: Nested Loops for Patterns
print("\n" + "=" * 60)
print("Program 6: Nested Loops for Patterns")
print("=" * 60)

# Pattern 1: Right triangle with numbers
print("\nPattern 1: Right Triangle (Numbers):")
print("-" * 30)
n = int(input("Enter the height: "))
for i in range(1, n + 1):
    for j in range(1, i + 1):
        print(j, end=' ')
    print()

# Pattern 2: Right triangle with stars
print("\nPattern 2: Right Triangle (Stars):")
print("-" * 30)
for i in range(1, n + 1):
    for j in range(1, i + 1):
        print('*', end=' ')
    print()

# Pattern 3: Pyramid
print("\nPattern 3: Pyramid:")
print("-" * 30)
for i in range(1, n + 1):
    for j in range(n - i):
        print(' ', end=' ')
    for j in range(i):
        print('*', end=' ')
    print()

# Pattern 4: Multiplication grid
print("\nPattern 4: Multiplication Grid (3x3):")
print("-" * 30)
for i in range(1, 4):
    for j in range(1, 4):
        print(f"{i*j:>3}", end=' ')
    print()

# Pattern 5: Diamond pattern
print("\nPattern 5: Diamond Pattern:")
print("-" * 30)
rows = 5
for i in range(rows):
    for j in range(rows - i - 1):
        print(' ', end=' ')
    for j in range(i + 1):
        print('*', end=' ')
    print()

for i in range(rows - 2, -1, -1):
    for j in range(rows - i - 1):
        print(' ', end=' ')
    for j in range(i + 1):
        print('*', end=' ')
    print()
```

## Part 7: break and continue Statements

```python
# Program 7: break and continue Statements
print("\n" + "=" * 60)
print("Program 7: break and continue Statements")
print("=" * 60)

# 1. Using break statement
print("\n1. Using break statement (stop at 5):")
print("-" * 30)
for i in range(1, 11):
    if i == 5:
        print(f"Breaking at {i}")
        break
    print(i, end=' ')
print("\n")

# 2. Using continue statement
print("2. Using continue statement (skip even numbers):")
print("-" * 30)
for i in range(1, 11):
    if i % 2 == 0:
        continue  # Skip even numbers
    print(i, end=' ')
print("\n")

# 3. Find first number divisible by 7
print("3. Find first number divisible by 7 (from 1 to 100):")
print("-" * 30)
for i in range(1, 101):
    if i % 7 == 0:
        print(f"First number divisible by 7 is: {i}")
        break

# 4. Skip numbers divisible by 3
print("\n4. Print numbers 1 to 20 (skip multiples of 3):")
print("-" * 30)
for i in range(1, 21):
    if i % 3 == 0:
        continue
    print(i, end=' ')
print("\n")

# 5. Using break in nested loops
print("5. Break in nested loops:")
print("-" * 30)
for i in range(1, 4):
    print(f"Row {i}: ", end='')
    for j in range(1, 5):
        if j == 3:
            break
        print(j, end=' ')
    print()
```

## Part 8: Comparison of for Loop and while Loop

```python
# Program 8: Comparison of for and while loops
print("\n" + "=" * 60)
print("Program 8: Comparison of for and while Loops")
print("=" * 60)

print("\nTask: Print numbers from 1 to 5")
print("-" * 60)

# Using for loop
print("\nUsing FOR loop:")
print("Code:")
print("for i in range(1, 6):")
print("    print(i)")
print("\nOutput:")
for i in range(1, 6):
    print(i)

# Using while loop
print("\nUsing WHILE loop:")
print("Code:")
print("i = 1")
print("while i <= 5:")
print("    print(i)")
print("    i += 1")
print("\nOutput:")
i = 1
while i <= 5:
    print(i)
    i += 1

# Comparison table
print("\n" + "=" * 60)
print("Comparison Table: for vs while")
print("=" * 60)

comparison = [
    ["Feature", "for Loop", "while Loop"],
    ["-" * 15, "-" * 20, "-" * 20],
    ["Number of iterations", "Known/Fixed", "May be unknown"],
    ["Initialization", "Automatic", "Manual required"],
    ["Syntax", "for var in sequence:", "while condition:"],
    ["Used for", "Iterating lists/ranges", "Conditional loops"],
    ["Performance", "Generally faster", "Similar but more flexible"],
    ["Readability", "More readable", "Less readable"],
    ["Use case", "Fixed iterations", "Unknown/dynamic iterations"]
]

for row in comparison:
    print(f"{row[0]:<20} | {row[1]:<30} | {row[2]:<30}")

# Example: Interactive program with while
print("\n" + "=" * 60)
print("Example: Interactive Sum Calculator (while loop)")
print("=" * 60)
print("\nNote: This demonstrates while loop's advantage for interactive input")
sum_interactive = 0
while True:
    num = int(input("\nEnter a number (enter 0 to stop): "))
    if num == 0:
        break
    sum_interactive += num
    print(f"Sum so far: {sum_interactive}")

print(f"\nFinal sum: {sum_interactive}")
```

## Part 9: Practical Applications of Loops

```python
# Program 9: Practical Applications
print("\n" + "=" * 60)
print("Program 9: Practical Applications of Loops")
print("=" * 60)

# 1. Factorial calculation
print("\n1. Factorial Calculation:")
print("-" * 30)
num_fact = int(input("Enter a number to find factorial: "))
factorial = 1
for i in range(1, num_fact + 1):
    factorial *= i
print(f"Factorial of {num_fact} is: {factorial}")

# 2. Prime number check
print("\n2. Prime Number Checker:")
print("-" * 30)
num_prime = int(input("Enter a number: "))
is_prime = True
if num_prime > 1:
    for i in range(2, int(num_prime**0.5) + 1):
        if num_prime % i == 0:
            is_prime = False
            break
else:
    is_prime = False

if is_prime:
    print(f"{num_prime} is a PRIME number")
else:
    print(f"{num_prime} is NOT a prime number")

# 3. Pattern: Floyd's Triangle
print("\n3. Floyd's Triangle:")
print("-" * 30)
rows_floyd = int(input("Enter number of rows: "))
num_floyd = 1
for i in range(1, rows_floyd + 1):
    for j in range(i):
        print(num_floyd, end=' ')
        num_floyd += 1
    print()

# 4. Average of numbers
print("\n4. Average of N Numbers:")
print("-" * 30)
count = int(input("How many numbers? "))
total = 0
for i in range(count):
    num = float(input(f"Enter number {i+1}: "))
    total += num
average = total / count
print(f"\nAverage: {average:.2f}")
```

## Key Observations and Learning Summary

### Important Concepts Learned:

#### 1. **for Loop:**
   - **Syntax:** `for variable in sequence:`
   - Used when number of iterations is known
   - Iterates through list, tuple, range, string, etc.
   - Automatically handles counter increment
   - **Example:** `for i in range(1, 6):`

#### 2. **while Loop:**
   - **Syntax:** `while condition:`
   - Used when number of iterations is unknown
   - Continues until condition becomes False
   - Manual counter management required
   - Risk of infinite loops if condition never becomes False
   - **Example:** `while i <= 5:`

#### 3. **break Statement:**
   - Immediately exits/terminates the loop
   - Control passes to statement after the loop
   - Used to stop loop when specific condition is met
   - Works in both for and while loops

#### 4. **continue Statement:**
   - Skips current iteration
   - Goes directly to next iteration
   - Does not terminate the loop
   - Used to skip processing for certain values

#### 5. **Nested Loops:**
   - Loop inside another loop
   - Outer loop executes once, inner loop executes completely
   - Used for creating patterns and grids
   - Requires proper indentation

#### 6. **range() Function:**
   - `range(start, stop, step)`
   - `range(5)` → 0, 1, 2, 3, 4
   - `range(1, 5)` → 1, 2, 3, 4
   - `range(1, 10, 2)` → 1, 3, 5, 7, 9
   - `range(10, 0, -1)` → 10, 9, 8, ..., 1

### Comparison: for vs while

| Aspect | for Loop | while Loop |
|--------|----------|------------|
| **Iterations** | Known/Fixed | Unknown/Dynamic |
| **Use Case** | Iterating sequences | Conditional repetition |
| **Initialization** | Automatic | Manual |
| **Code Length** | Concise | More verbose |
| **Readability** | Better for fixed iterations | Better for conditions |
| **Performance** | Slightly faster | Similar |
| **Risk** | Less risk of infinite loops | Risk of infinite loops |

### Common Loop Applications:
1. **Iteration patterns** - sequences and ranges
2. **Sum/Product calculation** - aggregate values
3. **Searching** - find elements meeting criteria
4. **Validation** - check conditions for multiple values
5. **Pattern generation** - create visual patterns
6. **Interactive input** - repeat until condition met

### Best Practices:
- Use for loop when iteration count is known
- Use while loop for unknown/dynamic iterations
- Always initialize counters in while loops
- Use break to exit prematurely when needed
- Use continue to skip unwanted iterations
- Avoid infinite loops
- Keep proper indentation for nested loops
- Test with edge cases (0, negative numbers, empty sequences)
