## Part 1: Understanding if Statement

```python
# Program 1: Simple if Statement
print("=" * 60)
print("Program 1: Simple if Statement")
print("=" * 60)

# Taking input from user
age = int(input("\nEnter your age: "))

# Simple if statement
if age >= 18:
    print(f"You are {age} years old.")
    print("You are an adult and eligible to vote.")

print("\nProgram execution completed.")
```

## Part 2: Understanding if-else Statement

```python
# Program 2: if-else Statement
print("\n" + "=" * 60)
print("Program 2: if-else Statement")
print("=" * 60)

# Taking input from user
age = int(input("\nEnter your age: "))

# if-else statement
if age >= 18:
    print(f"You are {age} years old.")
    print("You are an adult.")
else:
    print(f"You are {age} years old.")
    print("You are a minor.")

print("\nProgram execution completed.")
```

## Part 3: Understanding if-elif-else Statement

```python
# Program 3: if-elif-else Statement
print("\n" + "=" * 60)
print("Program 3: if-elif-else Statement")
print("=" * 60)

# Taking input from user
age = int(input("\nEnter your age: "))

# if-elif-else statement
if age < 13:
    print(f"You are {age} years old.")
    print("You are a child.")
elif age < 18:
    print(f"You are {age} years old.")
    print("You are a teenager.")
elif age < 60:
    print(f"You are {age} years old.")
    print("You are an adult.")
else:
    print(f"You are {age} years old.")
    print("You are a senior citizen.")

print("\nProgram execution completed.")
```

## Part 4: Even-Odd Checker

```python
# Program 4: Even-Odd Checker
print("\n" + "=" * 60)
print("Program 4: Even-Odd Number Checker")
print("=" * 60)

# Taking input from user
number = int(input("\nEnter a number: "))

# Check if number is even or odd
if number % 2 == 0:
    print(f"\n{number} is an EVEN number.")
    print(f"Explanation: {number} % 2 = {number % 2}")
else:
    print(f"\n{number} is an ODD number.")
    print(f"Explanation: {number} % 2 = {number % 2}")

# Test with multiple values
print("\n" + "-" * 60)
print("Testing with predefined values:")
print("-" * 60)

test_numbers = [10, 15, 20, 25, 100, 99]
for num in test_numbers:
    if num % 2 == 0:
        print(f"{num} is EVEN")
    else:
        print(f"{num} is ODD")
```

## Part 5: Positive-Negative Checker

```python
# Program 5: Positive-Negative-Zero Checker
print("\n" + "=" * 60)
print("Program 5: Positive-Negative-Zero Checker")
print("=" * 60)

# Taking input from user
number = float(input("\nEnter a number: "))

# Check if number is positive, negative, or zero
if number > 0:
    print(f"\n{number} is a POSITIVE number.")
elif number < 0:
    print(f"\n{number} is a NEGATIVE number.")
else:
    print(f"\nThe number is ZERO.")

# Test with multiple values
print("\n" + "-" * 60)
print("Testing with predefined values:")
print("-" * 60)

test_numbers = [5, -5, 0, 100, -50, 0.5, -0.5]
for num in test_numbers:
    if num > 0:
        print(f"{num:>6} is POSITIVE")
    elif num < 0:
        print(f"{num:>6} is NEGATIVE")
    else:
        print(f"{num:>6} is ZERO")
```

## Part 6: Largest of Two Numbers

```python
# Program 6: Find Largest of Two Numbers
print("\n" + "=" * 60)
print("Program 6: Find Largest of Two Numbers")
print("=" * 60)

# Taking input from user
num1 = float(input("\nEnter first number: "))
num2 = float(input("Enter second number: "))

print(f"\nComparing {num1} and {num2}:")

# Find largest
if num1 > num2:
    print(f"\nLargest number is: {num1}")
elif num2 > num1:
    print(f"\nLargest number is: {num2}")
else:
    print(f"\nBoth numbers are equal: {num1}")

# Test with multiple pairs
print("\n" + "-" * 60)
print("Testing with predefined pairs:")
print("-" * 60)

test_pairs = [(10, 20), (50, 30), (15, 15), (100, 99), (-5, -10)]
for a, b in test_pairs:
    if a > b:
        print(f"({a}, {b}) -> Largest: {a}")
    elif b > a:
        print(f"({a}, {b}) -> Largest: {b}")
    else:
        print(f"({a}, {b}) -> Both are equal: {a}")
```

## Part 7: Largest of Three Numbers

```python
# Program 7: Find Largest of Three Numbers
print("\n" + "=" * 60)
print("Program 7: Find Largest of Three Numbers")
print("=" * 60)

# Taking input from user
num1 = float(input("\nEnter first number: "))
num2 = float(input("Enter second number: "))
num3 = float(input("Enter third number: "))

print(f"\nComparing {num1}, {num2}, and {num3}:")

# Method 1: Using nested if-else
if num1 >= num2 and num1 >= num3:
    largest = num1
elif num2 >= num1 and num2 >= num3:
    largest = num2
else:
    largest = num3

print(f"\nLargest number is: {largest}")

# Test with multiple triplets
print("\n" + "-" * 60)
print("Testing with predefined triplets:")
print("-" * 60)

test_triplets = [(10, 20, 15), (5, 5, 5), (100, 50, 75), (30, 10, 20), (-1, -5, -2)]
for a, b, c in test_triplets:
    if a >= b and a >= c:
        max_val = a
    elif b >= a and b >= c:
        max_val = b
    else:
        max_val = c
    print(f"({a}, {b}, {c}) -> Largest: {max_val}")
```

## Part 8: Grading System

```python
# Program 8: Student Grading System
print("\n" + "=" * 60)
print("Program 8: Student Grading System")
print("=" * 60)

# Taking input from user
student_name = input("\nEnter student name: ")
marks = float(input("Enter marks obtained (out of 100): "))

# Validate marks
if marks < 0 or marks > 100:
    print("\nError: Marks should be between 0 and 100!")
else:
    # Determine grade
    if marks >= 90:
        grade = 'A+'
        remarks = "Excellent"
    elif marks >= 80:
        grade = 'A'
        remarks = "Very Good"
    elif marks >= 70:
        grade = 'B'
        remarks = "Good"
    elif marks >= 60:
        grade = 'C'
        remarks = "Average"
    elif marks >= 50:
        grade = 'D'
        remarks = "Below Average"
    else:
        grade = 'F'
        remarks = "Fail"
    
    # Display result
    print(f"\nStudent Name: {student_name}")
    print(f"Marks Obtained: {marks}")
    print(f"Grade: {grade}")
    print(f"Remarks: {remarks}")

# Test with multiple students
print("\n" + "-" * 60)
print("Testing with predefined student records:")
print("-" * 60)

students = [
    ("Abhay Rana", 95),
    ("Raj Kumar", 82),
    ("Priya Singh", 75),
    ("Anil Sharma", 65),
    ("Sakshi Verma", 55),
    ("Ravi Patel", 35)
]

for name, marks_val in students:
    if marks_val >= 90:
        grade_val = 'A+'
    elif marks_val >= 80:
        grade_val = 'A'
    elif marks_val >= 70:
        grade_val = 'B'
    elif marks_val >= 60:
        grade_val = 'C'
    elif marks_val >= 50:
        grade_val = 'D'
    else:
        grade_val = 'F'
    
    print(f"{name:20} | Marks: {marks_val:>3} | Grade: {grade_val}")
```

## Part 9: Nested Conditionals - Triangle Classification

```python
# Program 9: Triangle Classification using Nested Conditionals
print("\n" + "=" * 60)
print("Program 9: Triangle Classification")
print("=" * 60)

# Taking input from user
a = float(input("\nEnter first side: "))
b = float(input("Enter second side: "))
c = float(input("Enter third side: "))

# Check if it forms a valid triangle
if a + b > c and b + c > a and a + c > b:
    # Valid triangle - classify
    if a == b == c:
        triangle_type = "Equilateral"
    elif a == b or b == c or a == c:
        triangle_type = "Isosceles"
    else:
        triangle_type = "Scalene"
    
    # Check if right-angled
    sides = sorted([a, b, c])
    if sides[0]**2 + sides[1]**2 == sides[2]**2:
        triangle_type += " (Right-angled)"
    
    print(f"\nThe triangle with sides {a}, {b}, {c} is: {triangle_type}")
else:
    print(f"\nSides {a}, {b}, {c} do NOT form a valid triangle.")
    print("Reason: Sum of any two sides must be greater than the third side.")
```

## Key Observations and Learning Summary

### Important Concepts Learned:

#### 1. **if Statement:**
   - Executes a block of code only if the condition is True
   - Syntax: `if condition:`
   - No alternative path if condition is False

#### 2. **if-else Statement:**
   - Executes one block if condition is True, another if False
   - Syntax: `if condition: ... else: ...`
   - Always executes one of the two blocks

#### 3. **if-elif-else Statement:**
   - Tests multiple conditions in sequence
   - Stops at the first True condition
   - Can have multiple elif blocks
   - Syntax: `if condition1: ... elif condition2: ... else: ...`

#### 4. **Nested Conditionals:**
   - Conditional statements inside other conditional statements
   - Useful for complex decision-making
   - Requires proper indentation

#### 5. **Important Points:**
   - **Indentation is crucial** - Python uses indentation to define blocks
   - Colon (:) is required after condition
   - Conditions return Boolean values (True/False)
   - Comparison operators: `>`, `<`, `>=`, `<=`, `==`, `!=`
   - Logical operators: `and`, `or`, `not`
