## Part 1: List Creation and Basic Operations

```python
# Program 1: List Creation and Basic Operations
print("=" * 70)
print("Program 1: List Creation and Basic Operations")
print("=" * 70)

# 1. Creating lists
print("\n1. CREATING LISTS:")
print("-" * 70)

# Empty list
empty_list = []
print(f"Empty list: {empty_list}")
print(f"Type: {type(empty_list).__name__}")
print(f"Length: {len(empty_list)}")

# List with elements
numbers = [1, 2, 3, 4, 5]
print(f"\nNumbers list: {numbers}")

# Mixed data types
mixed_list = [1, "Hello", 3.14, True, [1, 2, 3]]
print(f"Mixed list: {mixed_list}")

# List from range
range_list = list(range(1, 11))
print(f"Range list: {range_list}")

# List from string
string_list = list("PYTHON")
print(f"String to list: {string_list}")

# 2. Basic operations
print("\n2. BASIC OPERATIONS:")
print("-" * 70)

fruits = ["apple", "banana", "cherry"]
print(f"Original list: {fruits}")

# append() - Add element at end
fruits.append("date")
print(f"After append('date'): {fruits}")

# insert() - Add element at specific position
fruits.insert(1, "orange")
print(f"After insert(1, 'orange'): {fruits}")

# remove() - Remove specific element
fruits.remove("banana")
print(f"After remove('banana'): {fruits}")

# pop() - Remove and return last element
popped = fruits.pop()
print(f"After pop(): {fruits}")
print(f"Popped element: {popped}")

# pop() with index
popped_index = fruits.pop(1)
print(f"After pop(1): {fruits}")
print(f"Popped element at index 1: {popped_index}")

# extend() - Add multiple elements
fruits.extend(["grape", "kiwi"])
print(f"After extend(['grape', 'kiwi']): {fruits}")

# clear() - Remove all elements
fruits_copy = fruits.copy()
fruits_copy.clear()
print(f"After clear(): {fruits_copy}")
print(f"Original list unchanged: {fruits}")

# 3. Sorting
print("\n3. SORTING OPERATIONS:")
print("-" * 70)

numbers_unsorted = [5, 2, 8, 1, 9, 3]
print(f"Unsorted numbers: {numbers_unsorted}")

# sort() - Sort in place
numbers_unsorted.sort()
print(f"After sort(): {numbers_unsorted}")

# sort() in reverse
numbers_unsorted.sort(reverse=True)
print(f"After sort(reverse=True): {numbers_unsorted}")

# sorted() - Return sorted copy
original = [3, 1, 4, 1, 5, 9, 2, 6]
sorted_copy = sorted(original)
print(f"\nOriginal: {original}")
print(f"Sorted copy: {sorted_copy}")
print(f"Original unchanged: {original}")

# reverse() - Reverse in place
original.reverse()
print(f"After reverse(): {original}")

# Sorting strings
words = ["zebra", "apple", "Banana", "cherry"]
print(f"\nWords: {words}")
words.sort()
print(f"After sort(): {words}")
words.sort(key=str.lower)
print(f"After sort(key=str.lower): {words}")
```

## Part 2: List Indexing and Slicing

```python
# Program 2: List Indexing and Slicing
print("\n" + "=" * 70)
print("Program 2: List Indexing and Slicing")
print("=" * 70)

# Create a sample list
letters = ['A', 'B', 'C', 'D', 'E', 'F', 'G', 'H', 'I', 'J']
print(f"List: {letters}")
print(f"Length: {len(letters)}")

# 1. Indexing
print("\n1. INDEXING:")
print("-" * 70)
print(f"letters[0] = '{letters[0]}'  # First element")
print(f"letters[4] = '{letters[4]}'  # Fifth element")
print(f"letters[-1] = '{letters[-1]}'  # Last element")
print(f"letters[-3] = '{letters[-3]}'  # Third from last")

# 2. Slicing
print("\n2. SLICING:")
print("-" * 70)
print(f"letters[0:5] = {letters[0:5]}  # From index 0 to 4")
print(f"letters[2:7] = {letters[2:7]}  # From index 2 to 6")
print(f"letters[:5] = {letters[:5]}  # First 5 elements")
print(f"letters[5:] = {letters[5:]}  # From index 5 to end")
print(f"letters[::2] = {letters[::2]}  # Every 2nd element")
print(f"letters[1::2] = {letters[1::2]}  # Every 2nd element starting from index 1")
print(f"letters[::-1] = {letters[::-1]}  # Reverse the list")
print(f"letters[2:8:2] = {letters[2:8:2]}  # Every 2nd element from index 2 to 7")

# 3. Modifying elements using indexing
print("\n3. MODIFYING ELEMENTS:")
print("-" * 70)
numbers = [10, 20, 30, 40, 50]
print(f"Original: {numbers}")

# Change single element
numbers[2] = 35
print(f"After numbers[2] = 35: {numbers}")

# Change multiple elements using slicing
numbers[1:4] = [25, 35, 45]
print(f"After numbers[1:4] = [25, 35, 45]: {numbers}")

# Delete elements using slicing
numbers[2:4] = []
print(f"After numbers[2:4] = []: {numbers}")

# 4. List membership
print("\n4. MEMBERSHIP TESTING:")
print("-" * 70)
fruits = ['apple', 'banana', 'cherry', 'date']
print(f"List: {fruits}")
print(f"'apple' in fruits: {'apple' in fruits}")
print(f"'grape' in fruits: {'grape' in fruits}")
print(f"'apple' not in fruits: {'apple' not in fruits}")

# 5. Finding elements
print("\n5. FINDING ELEMENTS:")
print("-" * 70)
print(f"fruits.index('cherry'): {fruits.index('cherry')}")
print(f"fruits.count('apple'): {fruits.count('apple')}")
print(f"fruits.count('grape'): {fruits.count('grape')}")

# 6. List concatenation and repetition
print("\n6. CONCATENATION AND REPETITION:")
print("-" * 70)
list1 = [1, 2, 3]
list2 = [4, 5, 6]
print(f"list1: {list1}")
print(f"list2: {list2}")
print(f"list1 + list2: {list1 + list2}")
print(f"list1 * 3: {list1 * 3}")
print(f"list2 * 2: {list2 * 2}")
```

## Part 3: Tuple Creation and Properties

```python
# Program 3: Tuple Creation and Properties
print("\n" + "=" * 70)
print("Program 3: Tuple Creation and Properties")
print("=" * 70)

# 1. Creating tuples
print("\n1. CREATING TUPLES:")
print("-" * 70)

# Empty tuple
empty_tuple = ()
print(f"Empty tuple: {empty_tuple}")
print(f"Type: {type(empty_tuple).__name__}")
print(f"Length: {len(empty_tuple)}")

# Tuple with elements
numbers_tuple = (1, 2, 3, 4, 5)
print(f"\nNumbers tuple: {numbers_tuple}")

# Mixed data types
mixed_tuple = (1, "Hello", 3.14, True, [1, 2, 3])
print(f"Mixed tuple: {mixed_tuple}")

# Single element tuple (note the comma)
single_tuple = (42,)
print(f"Single element tuple: {single_tuple}")
print(f"Type: {type(single_tuple)}")

# Tuple without parentheses
no_parens = 1, 2, 3, 4
print(f"Tuple without parentheses: {no_parens}")

# Tuple from list
list_to_tuple = tuple([1, 2, 3, 4, 5])
print(f"Tuple from list: {list_to_tuple}")

# 2. Accessing tuple elements
print("\n2. ACCESSING TUPLE ELEMENTS:")
print("-" * 70)
colors = ('red', 'green', 'blue', 'yellow', 'purple')
print(f"Tuple: {colors}")
print(f"colors[0]: {colors[0]}")
print(f"colors[2]: {colors[2]}")
print(f"colors[-1]: {colors[-1]}")
print(f"colors[-2]: {colors[-2]}")

# Slicing
print(f"\ncolors[1:4]: {colors[1:4]}")
print(f"colors[:3]: {colors[:3]}")
print(f"colors[2:]: {colors[2:]}")
print(f"colors[::2]: {colors[::2]}")
print(f"colors[::-1]: {colors[::-1]}")

# 3. Tuple immutability
print("\n3. TUPLE IMMUTABILITY:")
print("-" * 70)
print("Tuples are IMMUTABLE - their elements cannot be changed after creation")

coordinates = (10, 20, 30)
print(f"Original tuple: {coordinates}")

# This will work - accessing elements
print(f"coordinates[1]: {coordinates[1]}")

# This will NOT work - trying to modify
try:
    coordinates[1] = 25
except TypeError as e:
    print(f"Error when trying to modify: {e}")

# This will NOT work - trying to append
try:
    coordinates.append(40)
except AttributeError as e:
    print(f"Error when trying to append: {e}")

# 4. Tuple operations that work
print("\n4. OPERATIONS THAT WORK WITH TUPLES:")
print("-" * 70)

tuple1 = (1, 2, 3)
tuple2 = (4, 5, 6)
print(f"tuple1: {tuple1}")
print(f"tuple2: {tuple2}")

# Concatenation
print(f"tuple1 + tuple2: {tuple1 + tuple2}")

# Repetition
print(f"tuple1 * 3: {tuple1 * 3}")

# Membership
print(f"2 in tuple1: {2 in tuple1}")
print(f"7 in tuple1: {7 in tuple1}")

# Length
print(f"len(tuple1): {len(tuple1)}")

# Count and index
print(f"tuple1.count(2): {tuple1.count(2)}")
print(f"tuple1.index(3): {tuple1.index(3)}")

# 5. Tuple unpacking
print("\n5. TUPLE UNPACKING:")
print("-" * 70)
person = ('John', 25, 'Engineer')
print(f"Tuple: {person}")

# Unpack into variables
name, age, profession = person
print(f"Name: {name}")
print(f"Age: {age}")
print(f"Profession: {profession}")

# Partial unpacking
first, *rest = person
print(f"\nFirst: {first}")
print(f"Rest: {rest}")

# 6. Nested tuples
print("\n6. NESTED TUPLES:")
print("-" * 70)
nested = ((1, 2), (3, 4), (5, 6))
print(f"Nested tuple: {nested}")
print(f"nested[0]: {nested[0]}")
print(f"nested[0][1]: {nested[0][1]}")
print(f"nested[1:3]: {nested[1:3]}")
```

## Part 4: Dictionary Creation and Operations

```python
# Program 4: Dictionary Creation and Operations
print("\n" + "=" * 70)
print("Program 4: Dictionary Creation and Operations")
print("=" * 70)

# 1. Creating dictionaries
print("\n1. CREATING DICTIONARIES:")
print("-" * 70)

# Empty dictionary
empty_dict = {}
print(f"Empty dictionary: {empty_dict}")
print(f"Type: {type(empty_dict).__name__}")
print(f"Length: {len(empty_dict)}")

# Dictionary with key-value pairs
student = {
    'name': 'Alice',
    'age': 20,
    'grade': 'A',
    'marks': 95
}
print(f"\nStudent dictionary: {student}")

# Using dict() constructor
dict_constructor = dict(name='Bob', age=22, city='New York')
print(f"Using dict() constructor: {dict_constructor}")

# From list of tuples
pairs = [('x', 1), ('y', 2), ('z', 3)]
dict_from_pairs = dict(pairs)
print(f"From list of tuples: {dict_from_pairs}")

# 2. Accessing dictionary values
print("\n2. ACCESSING DICTIONARY VALUES:")
print("-" * 70)
print(f"student['name']: {student['name']}")
print(f"student['age']: {student['age']}")
print(f"student.get('grade'): {student.get('grade')}")
print(f"student.get('city', 'Not found'): {student.get('city', 'Not found')}")

# 3. Adding and updating items
print("\n3. ADDING AND UPDATING ITEMS:")
print("-" * 70)
print(f"Original: {student}")

# Add new key-value pair
student['city'] = 'Boston'
print(f"After student['city'] = 'Boston': {student}")

# Update existing value
student['marks'] = 97
print(f"After student['marks'] = 97: {student}")

# Using update() method
student.update({'grade': 'A+', 'subject': 'Math'})
print(f"After update(): {student}")

# 4. Removing items
print("\n4. REMOVING ITEMS:")
print("-" * 70)
print(f"Current: {student}")

# Using del
if 'subject' in student:
    del student['subject']
print(f"After del student['subject']: {student}")

# Using pop()
removed_value = student.pop('city')
print(f"After pop('city'): {student}")
print(f"Removed value: {removed_value}")

# Using pop() with default
removed_default = student.pop('address', 'Not found')
print(f"pop('address', 'Not found'): {removed_default}")

# Using popitem() - removes last item
last_item = student.popitem()
print(f"After popitem(): {student}")
print(f"Removed item: {last_item}")

# 5. Dictionary methods
print("\n5. DICTIONARY METHODS:")
print("-" * 70)
sample_dict = {'a': 1, 'b': 2, 'c': 3, 'd': 4}
print(f"Sample dictionary: {sample_dict}")

# Keys
print(f"Keys: {list(sample_dict.keys())}")

# Values
print(f"Values: {list(sample_dict.values())}")

# Items (key-value pairs)
print(f"Items: {list(sample_dict.items())}")

# Membership
print(f"'a' in sample_dict: {'a' in sample_dict}")
print(f"'z' in sample_dict: {'z' in sample_dict}")
print(f"'x' not in sample_dict: {'x' not in sample_dict}")

# Copy
copied_dict = sample_dict.copy()
print(f"Copied dictionary: {copied_dict}")

# Clear
copied_dict.clear()
print(f"After clear(): {copied_dict}")
print(f"Original unchanged: {sample_dict}")

# 6. Dictionary comprehension
print("\n6. DICTIONARY COMPREHENSION:")
print("-" * 70)

# Create dictionary from range
squares = {x: x**2 for x in range(1, 6)}
print(f"Squares: {squares}")

# Create dictionary from two lists
keys = ['name', 'age', 'city']
values = ['Alice', 25, 'NYC']
combined = {keys[i]: values[i] for i in range(len(keys))}
print(f"Combined: {combined}")

# Filter dictionary
filtered = {k: v for k, v in squares.items() if v > 10}
print(f"Filtered (values > 10): {filtered}")
```

## Part 5: Traversing Collections with Loops

```python
# Program 5: Traversing Collections with Loops
print("\n" + "=" * 70)
print("Program 5: Traversing Collections with Loops")
print("=" * 70)

# Sample data
fruits_list = ['apple', 'banana', 'cherry', 'date', 'elderberry']
colors_tuple = ('red', 'green', 'blue', 'yellow', 'purple')
student_dict = {
    'name': 'Alice',
    'age': 20,
    'grade': 'A',
    'marks': 95,
    'city': 'Boston'
}

# 1. Traversing a list
print("\n1. TRAVERSING A LIST:")
print("-" * 70)
print(f"List: {fruits_list}")
print("\nUsing for loop:")
for fruit in fruits_list:
    print(f"Fruit: {fruit}")

print("Using for loop with index:")
for i, fruit in enumerate(fruits_list):
    print(f"Index {i}: {fruit}")

print("Using while loop:")
i = 0
while i < len(fruits_list):
    print(f"Fruit {i+1}: {fruits_list[i]}")
    i += 1

# 2. Traversing a tuple
print("\n2. TRAVERSING A TUPLE:")
print("-" * 70)
print(f"Tuple: {colors_tuple}")
print("\nUsing for loop:")
for color in colors_tuple:
    print(f"Color: {color}")

print("Using for loop with index:")
for i, color in enumerate(colors_tuple):
    print(f"Index {i}: {color}")

print("Using while loop:")
i = 0
while i < len(colors_tuple):
    print(f"Color {i+1}: {colors_tuple[i]}")
    i += 1

# 3. Traversing a dictionary
print("\n3. TRAVERSING A DICTIONARY:")
print("-" * 70)
print(f"Dictionary: {student_dict}")

print("Iterating over keys:")
for key in student_dict:
    print(f"Key: {key}")

print("\nIterating over keys explicitly:")
for key in student_dict.keys():
    print(f"Key: {key}")

print("\nIterating over values:")
for value in student_dict.values():
    print(f"Value: {value}")

print("\nIterating over key-value pairs:")
for key, value in student_dict.items():
    print(f"{key}: {value}")

print("\nIterating with enumeration:")
for i, (key, value) in enumerate(student_dict.items()):
    print(f"{i+1}. {key}: {value}")

# 4. Advanced traversal examples
print("\n4. ADVANCED TRAVERSAL EXAMPLES:")
print("-" * 70)

# List of dictionaries
students = [
    {'name': 'Alice', 'age': 20, 'grade': 'A'},
    {'name': 'Bob', 'age': 22, 'grade': 'B'},
    {'name': 'Charlie', 'age': 21, 'grade': 'A'}
]

print("List of dictionaries:")
for i, student in enumerate(students, 1):
    print(f"\nStudent {i}:")
    for key, value in student.items():
        print(f"  {key}: {value}")

# Nested list
matrix = [
    [1, 2, 3],
    [4, 5, 6],
    [7, 8, 9]
]

print("\nMatrix traversal:")
for i, row in enumerate(matrix):
    print(f"Row {i}: ", end='')
    for element in row:
        print(element, end=' ')
    print()

# Dictionary of lists
class_grades = {
    'Math': [85, 92, 78, 96],
    'Science': [88, 79, 91, 84],
    'English': [90, 87, 93, 89]
}

print("\nDictionary of lists:")
for subject, grades in class_grades.items():
    print(f"\n{subject}:")
    for grade in grades:
        print(f"  Grade: {grade}")
    print(f"  Average: {sum(grades)/len(grades):.1f}")
```

## Part 6: Comparing List, Tuple, and Dictionary

```python
# Program 6: Comparing List, Tuple, and Dictionary
print("\n" + "=" * 70)
print("Program 6: Comparing List, Tuple, and Dictionary")
print("=" * 70)

# Create sample data structures
sample_list = [1, 2, 3, 4, 5]
sample_tuple = (1, 2, 3, 4, 5)
sample_dict = {'a': 1, 'b': 2, 'c': 3, 'd': 4, 'e': 5}

print(f"List: {sample_list}")
print(f"Tuple: {sample_tuple}")
print(f"Dictionary: {sample_dict}")

# Detailed comparison table
comparison_data = [
    ["Feature", "List", "Tuple", "Dictionary"],
    ["-" * 15, "-" * 20, "-" * 20, "-" * 20],
    ["Syntax", "[1, 2, 3]", "(1, 2, 3)", "{'a': 1, 'b': 2}"],
    ["Mutable", "Yes", "No", "Yes"],
    ["Ordered", "Yes", "Yes", "Python 3.7+"],
    ["Indexing", "Yes", "Yes", "By keys"],
    ["Slicing", "Yes", "Yes", "No"],
    ["Duplicates", "Allowed", "Allowed", "Keys: No, Values: Yes"],
    ["Methods", "Many", "Few", "Many"],
    ["Memory", "More", "Less", "More"],
    ["Speed", "Slower", "Faster", "Fast lookup"],
    ["Use Case", "Sequences", "Constants", "Key-value pairs"],
    ["Modification", "append, remove", "Immutable", "add, update, delete"],
    ["Iteration", "for item in list", "for item in tuple", "for key in dict"],
    ["Access", "list[0]", "tuple[0]", "dict['key']"],
    ["Length", "len(list)", "len(tuple)", "len(dict)"],
    ["Empty", "[]", "()", "{}" or dict()"],
    ["Type", "list", "tuple", "dict"],
]

print("\n" + "=" * 70)
print("DETAILED COMPARISON:")
print("=" * 70)
for row in comparison_data:
    print(f"{row[0]:<15} | {row[1]:<20} | {row[2]:<20} | {row[3]:<20}")

# Practical demonstration
print("\n" + "=" * 70)
print("PRACTICAL DEMONSTRATION:")
print("=" * 70)

# 1. Modifying elements
print("\n1. MODIFYING ELEMENTS:")
print("-" * 70)

# List - can modify
list_mod = [1, 2, 3]
print(f"Original list: {list_mod}")
list_mod[1] = 20
print(f"After modification: {list_mod}")

# Tuple - cannot modify
try:
    tuple_mod = (1, 2, 3)
    print(f"Original tuple: {tuple_mod}")
    tuple_mod[1] = 20
except TypeError as e:
    print(f"Tuple modification error: {e}")

# Dictionary - can modify
dict_mod = {'a': 1, 'b': 2, 'c': 3}
print(f"Original dict: {dict_mod}")
dict_mod['b'] = 20
dict_mod['d'] = 4
print(f"After modification: {dict_mod}")

# 2. Access methods
print("\n2. ACCESS METHODS:")
print("-" * 70)

# List/Tuple - by index
print(f"list_mod[0]: {list_mod[0]}")
print(f"tuple_mod[0]: {tuple_mod[0]}")

# Dictionary - by key
print(f"dict_mod['a']: {dict_mod['a']}")
print(f"dict_mod.get('z', 'Not found'): {dict_mod.get('z', 'Not found')}")

# 3. Performance comparison
print("\n3. PERFORMANCE COMPARISON:")
print("-" * 70)
import time

# Large collections
large_list = list(range(10000))
large_tuple = tuple(range(10000))
large_dict = {i: i for i in range(10000)}

# Test lookup time
# List lookup (worst case)
start = time.time()
result_list = 9999 in large_list
list_time = time.time() - start

# Tuple lookup
start = time.time()
result_tuple = 9999 in large_tuple
tuple_time = time.time() - start

# Dictionary lookup
start = time.time()
result_dict = 9999 in large_dict
dict_time = time.time() - start

print(f"List lookup (9999 in list): {list_time:.6f} seconds")
print(f"Tuple lookup (9999 in tuple): {tuple_time:.6f} seconds")
print(f"Dict lookup (9999 in dict): {dict_time:.6f} seconds")
print(f"\nDictionary is {list_time/dict_time:.1f}x faster than list for lookups!")

# 4. Memory usage
print("\n4. MEMORY USAGE:")
print("-" * 70)
import sys
print(f"List memory: {sys.getsizeof(large_list)} bytes")
print(f"Tuple memory: {sys.getsizeof(large_tuple)} bytes")
print(f"Dict memory: {sys.getsizeof(large_dict)} bytes")
print("\nTuples use less memory than lists, dictionaries use more for key storage.")
```

## Part 7: Practical Applications

```python
# Program 7: Practical Applications
print("\n" + "=" * 70)
print("Program 7: Practical Applications")
print("=" * 70)

# 1. Student Grade Management System
print("\n1. STUDENT GRADE MANAGEMENT:")
print("-" * 70)

# Using dictionary for student records
students_db = {
    'Alice': {'math': 95, 'science': 88, 'english': 92},
    'Bob': {'math': 78, 'science': 85, 'english': 80},
    'Charlie': {'math': 92, 'science': 91, 'english': 89}
}

# Calculate averages
for name, grades in students_db.items():
    avg = sum(grades.values()) / len(grades)
    print(f"{name}: Average = {avg:.1f}")

# 2. Shopping Cart
print("\n2. SHOPPING CART:")
print("-" * 70)

# Using list for cart items
cart = []
prices = {'apple': 2.5, 'banana': 1.5, 'orange': 3.0}

# Add items to cart
cart.extend(['apple', 'banana', 'apple', 'orange'])
print(f"Cart items: {cart}")

# Calculate total
cart_set = set(cart)  # Remove duplicates for pricing
total = sum(prices[item] for item in cart_set)
print(f"Total price: ${total:.2f}")

# 3. Configuration Settings
print("\n3. CONFIGURATION SETTINGS:")
print("-" * 70)

# Using tuple for immutable settings
app_config = (
    ('version', '1.0.0'),
    ('debug', False),
    ('max_users', 100),
    ('timeout', 30)
)

print("Application Configuration:")
for key, value in app_config:
    print(f"  {key}: {value}")

# 4. Word Frequency Counter
print("\n4. WORD FREQUENCY COUNTER:")
print("-" * 70)

text = "python is great python is powerful python is easy to learn"
words = text.split()
word_freq = {}

for word in words:
    word_freq[word] = word_freq.get(word, 0) + 1

print(f"Text: {text}")
print("Word frequencies:")
for word, count in sorted(word_freq.items()):
    print(f"  '{word}': {count}")

# 5. Matrix Operations
print("\n5. MATRIX OPERATIONS:")
print("-" * 70)

# Using list of lists for matrix
matrix = [
    [1, 2, 3],
    [4, 5, 6],
    [7, 8, 9]
]

print("Matrix:")
for row in matrix:
    print(f"  {row}")

# Calculate diagonal sum
diagonal_sum = sum(matrix[i][i] for i in range(len(matrix)))
print(f"Diagonal sum: {diagonal_sum}")

# 6. Phone Book
print("\n6. PHONE BOOK:")
print("-" * 70)

# Using dictionary for phone book
phone_book = {
    'Alice': '555-0101',
    'Bob': '555-0102',
    'Charlie': '555-0103'
}

# Search for contact
search_name = 'Bob'
if search_name in phone_book:
    print(f"{search_name}'s number: {phone_book[search_name]}")
else:
    print(f"{search_name} not found in phone book")

# Add new contact
phone_book['David'] = '555-0104'
print(f"Added David: {phone_book['David']}")
print(f"Total contacts: {len(phone_book)}")

# 7. Data Analysis
print("\n7. DATA ANALYSIS:")
print("-" * 70)

# Using tuple for immutable data points
data_points = (
    (25, 180, 75),  # (age, height, weight)
    (30, 175, 80),
    (35, 170, 70)
)

print("Data points (age, height, weight):")
for i, (age, height, weight) in enumerate(data_points, 1):
    print(f"  Person {i}: Age={age}, Height={height}cm, Weight={weight}kg")

# Calculate averages
ages = [point[0] for point in data_points]
heights = [point[1] for point in data_points]
weights = [point[2] for point in data_points]

print(f"\nAverage age: {sum(ages)/len(ages):.1f}")
print(f"Average height: {sum(heights)/len(heights):.1f} cm")
print(f"Average weight: {sum(weights)/len(weights):.1f} kg")
```

## Key Observations and Learning Summary

### LISTS - Key Points:

#### 1. **List Creation:**
   - `[]` or `list()` for empty list
   - `[1, 2, 3]` for list with elements
   - `list(range(5))` from range
   - `list("hello")` from string
   - Lists are **MUTABLE** (can be changed)

#### 2. **List Operations:**
   - **Adding:** `append()`, `insert()`, `extend()`
   - **Removing:** `remove()`, `pop()`, `clear()`
   - **Sorting:** `sort()`, `sorted()`
   - **Access:** `list[index]`, slicing `list[start:end]`
   - **Search:** `index()`, `count()`, membership with `in`

#### 3. **List Characteristics:**
   - **Ordered:** Elements maintain insertion order
   - **Indexed:** Access by position (0-based)
   - **Allows duplicates:** Same value can appear multiple times
   - **Dynamic size:** Can grow and shrink
   - **Heterogeneous:** Can contain different data types

---

### TUPLES - Key Points:

#### 1. **Tuple Creation:**
   - `()` or `tuple()` for empty tuple
   - `(1, 2, 3)` for tuple with elements
   - `(42,)` for single element (note comma)
   - `1, 2, 3` without parentheses
   - Tuples are **IMMUTABLE** (cannot be changed)

#### 2. **Tuple Operations:**
   - **Access:** `tuple[index]`, slicing `tuple[start:end]`
   - **Search:** `index()`, `count()`, membership with `in`
   - **Concatenation:** `tuple1 + tuple2`
   - **Repetition:** `tuple * 3`
   - **Unpacking:** `a, b, c = tuple`

#### 3. **Tuple Characteristics:**
   - **Ordered:** Elements maintain insertion order
   - **Indexed:** Access by position (0-based)
   - **Allows duplicates:** Same value can appear multiple times
   - **Fixed size:** Cannot change size after creation
   - **Heterogeneous:** Can contain different data types
   - **Hashable:** Can be used as dictionary keys

---

### DICTIONARIES - Key Points:

#### 1. **Dictionary Creation:**
   - `{}` or `dict()` for empty dictionary
   - `{'key': 'value'}` for dictionary with elements
   - `dict(key='value')` using constructor
   - `dict([('key', 'value')])` from list of tuples
   - Dictionaries are **MUTABLE** (can be changed)

#### 2. **Dictionary Operations:**
   - **Adding/Updating:** `dict['key'] = value`, `update()`
   - **Removing:** `del dict['key']`, `pop()`, `popitem()`, `clear()`
   - **Access:** `dict['key']`, `dict.get('key', default)`
   - **Views:** `keys()`, `values()`, `items()`
   - **Membership:** `'key' in dict`

#### 3. **Dictionary Characteristics:**
   - **Unordered (Python < 3.7):** No guaranteed order
   - **Ordered (Python 3.7+):** Maintains insertion order
   - **Key-based access:** Access by key, not position
   - **Unique keys:** Keys must be unique and hashable
   - **Duplicate values allowed:** Values can be duplicated
   - **Heterogeneous:** Keys and values can be different types

---

### COMPARISON SUMMARY:

| Feature | List | Tuple | Dictionary |
|---------|------|-------|------------|
| **Syntax** | `[1, 2, 3]` | `(1, 2, 3)` | `{'a': 1, 'b': 2}` |
| **Mutable** | Yes | No | Yes |
| **Ordered** | Yes | Yes | Python 3.7+ |
| **Access** | `list[0]` | `tuple[0]` | `dict['key']` |
| **Duplicates** | Allowed | Allowed | Keys: No, Values: Yes |
| **Use Case** | Sequences | Constants | Key-value mapping |
| **Memory** | More | Less | More |
| **Performance** | Good | Better | Best for lookups |

---

### PRACTICAL APPLICATIONS:

1. **Lists:** Shopping carts, task lists, data sequences, dynamic collections
2. **Tuples:** Coordinates, RGB colors, database records, function returns
3. **Dictionaries:** User profiles, configuration settings, lookup tables, JSON data

### CONCLUSION:

This experiment successfully demonstrated the three fundamental collection data types in Python:

1. **Lists** are versatile, mutable sequences ideal for dynamic data that needs frequent modification
2. **Tuples** are immutable, memory-efficient sequences perfect for constant data and hashable keys
3. **Dictionaries** provide fast key-based access and are essential for data mapping and lookup operations

Understanding when to use each data type is crucial for writing efficient and maintainable Python code. Each has its strengths and is designed for specific use cases in data storage and manipulation.
