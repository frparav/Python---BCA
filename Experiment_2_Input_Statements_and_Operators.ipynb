## Part 1: Taking User Input and Type Conversion

```python
# Program 1: Simple input and type conversion
print("=" * 50)
print("Program 1: User Input and Type Conversion")
print("=" * 50)

# Taking input from user
name = input("Enter your name: ")
age = int(input("Enter your age: "))
height = float(input("Enter your height (in meters): "))
marks = float(input("Enter your marks: "))

# Displaying the input values
print("\nYour Information:")
print(f"Name: {name}")
print(f"Age: {age} (Type: {type(age).__name__})")
print(f"Height: {height} (Type: {type(height).__name__})")
print(f"Marks: {marks} (Type: {type(marks).__name__})")
```

## Part 2: Arithmetic Operators

```python
# Program 2: Arithmetic Operations
print("\n" + "=" * 50)
print("Program 2: Arithmetic Operators")
print("=" * 50)

# Taking two numbers as input
num1 = float(input("\nEnter first number: "))
num2 = float(input("Enter second number: "))

print("\nArithmetic Operations:")
print(f"Number 1: {num1}")
print(f"Number 2: {num2}")
print()

# Addition
addition = num1 + num2
print(f"Addition ({num1} + {num2}): {addition}")

# Subtraction
subtraction = num1 - num2
print(f"Subtraction ({num1} - {num2}): {subtraction}")

# Multiplication
multiplication = num1 * num2
print(f"Multiplication ({num1} * {num2}): {multiplication}")

# Division
if num2 != 0:
    division = num1 / num2
    print(f"Division ({num1} / {num2}): {division}")
else:
    print("Division: Cannot divide by zero!")

# Floor Division
if num2 != 0:
    floor_division = num1 // num2
    print(f"Floor Division ({num1} // {num2}): {floor_division}")
else:
    print("Floor Division: Cannot divide by zero!")

# Modulus (Remainder)
if num2 != 0:
    modulus = num1 % num2
    print(f"Modulus ({num1} % {num2}): {modulus}")
else:
    print("Modulus: Cannot divide by zero!")

# Power
power = num1 ** num2
print(f"Power ({num1} ** {num2}): {power}")
```

## Part 3: Comparison Operators

```python
# Program 3: Comparison Operators
print("\n" + "=" * 50)
print("Program 3: Comparison Operators")
print("=" * 50)

# Taking input
a = int(input("\nEnter first number: "))
b = int(input("Enter second number: "))

print(f"\nComparing {a} and {b}:")
print()

# Greater than
print(f"{a} > {b}: {a > b}")

# Less than
print(f"{a} < {b}: {a < b}")

# Equal to
print(f"{a} == {b}: {a == b}")

# Not equal to
print(f"{a} != {b}: {a != b}")

# Greater than or equal to
print(f"{a} >= {b}: {a >= b}")

# Less than or equal to
print(f"{a} <= {b}: {a <= b}")
```

## Part 4: Logical Operators

```python
# Program 4: Logical Operators
print("\n" + "=" * 50)
print("Program 4: Logical Operators")
print("=" * 50)

# Taking input
x = int(input("\nEnter a number: "))

print(f"\nChecking if {x} is within range [10, 20]:")
print()

# AND operator
result_and = (x >= 10) and (x <= 20)
print(f"({x} >= 10) and ({x} <= 20): {result_and}")

# OR operator
result_or = (x < 10) or (x > 20)
print(f"({x} < 10) or ({x} > 20): {result_or}")

# NOT operator
result_not = not (x == 15)
print(f"not ({x} == 15): {result_not}")

print("\n" + "-" * 50)
print("Additional Logical Examples:")
print("-" * 50)

age = int(input("\nEnter your age: "))
has_license = input("Do you have a license? (yes/no): ").lower() == 'yes'

can_drive = (age >= 18) and has_license
print(f"\nCan drive (age >= 18 and has license): {can_drive}")

is_student_or_senior = (age < 25) or (age > 60)
print(f"Is student or senior (age < 25 or age > 60): {is_student_or_senior}")

not_valid = not can_drive
print(f"Cannot drive: {not_valid}")
```

## Part 5: Complete Calculator Program

```python
# Program 5: Complete Calculator Program using all operators
print("\n" + "=" * 50)
print("Program 5: Complete Calculator")
print("=" * 50)

print("\nSimple Calculator")
print("-" * 30)

# Input
num1 = float(input("Enter first number: "))
operator = input("Enter operator (+, -, *, /, %, **): ")
num2 = float(input("Enter second number: "))

# Calculation
print(f"\nCalculation: {num1} {operator} {num2}")

if operator == '+':
    result = num1 + num2
elif operator == '-':
    result = num1 - num2
elif operator == '*':
    result = num1 * num2
elif operator == '/':
    if num2 != 0:
        result = num1 / num2
    else:
        print("Error: Cannot divide by zero!")
        result = None
elif operator == '%':
    if num2 != 0:
        result = num1 % num2
    else:
        print("Error: Cannot divide by zero!")
        result = None
elif operator == '**':
    result = num1 ** num2
else:
    print("Invalid operator!")
    result = None

if result is not None:
    print(f"Result: {result}")
    # Validation
    print(f"\nResult is positive: {result > 0}")
    print(f"Result is zero: {result == 0}")
    print(f"Result is negative: {result < 0}")
```

## Observations and Learning Summary

### Key Observations:

1. **Input Function:**
   - The `input()` function always returns a string
   - Type conversion is necessary to perform arithmetic operations
   - `int()` converts to integer, `float()` converts to floating-point numbers

2. **Arithmetic Operators:**
   - `+` : Addition
   - `-` : Subtraction
   - `*` : Multiplication
   - `/` : Division (returns float)
   - `//` : Floor division (returns integer)
   - `%` : Modulus (remainder)
   - `**` : Power/Exponentiation

3. **Comparison Operators:**
   - Return Boolean values (True or False)
   - Used in conditional statements
   - Operators: `>`, `<`, `==`, `!=`, `>=`, `<=`

4. **Logical Operators:**
   - `and` : Both conditions must be True
   - `or` : At least one condition must be True
   - `not` : Reverses the Boolean value
   - Used to combine multiple conditions

5. **Error Handling:**
   - Always check for division by zero
   - Validate user input before performing operations
   - Use conditional statements to handle exceptions

6. **Best Practices:**
   - Always specify the data type during input conversion
   - Use meaningful variable names
   - Display clear output messages
   - Test programs with multiple inputs (boundary values, negative numbers, zero)
