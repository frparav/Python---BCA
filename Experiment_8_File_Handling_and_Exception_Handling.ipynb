```python
# Experiment 8: File Handling and Exception Handling
# Name: Abhay Rana
# SAP ID: 590024916

print("=" * 70)
print("EXPERIMENT 8: FILE HANDLING AND EXCEPTION HANDLING")
print("Name: Abhay Rana")
print("SAP ID: 590024916")
print("=" * 70)

print("\nParagraph Explanation:")
print(
    "This experiment explains the use of file handling and exception handling in "
    "Python. In this program, different text files are opened and processed to "
    "perform tasks such as counting names, finding the longest name, calculating "
    "maximum and average values, and reading city details. The experiment also "
    "shows how exceptions help in handling errors like missing files, invalid "
    "data, division by zero, and wrong user input. Custom exceptions are also "
    "used to make the program more clear and user friendly."
)


class FileHandlingError(Exception):
    """Base class for custom file handling exceptions."""


class EmptyFileError(FileHandlingError):
    """Raised when a required file is empty."""


class InvalidNumberError(FileHandlingError):
    """Raised when a file contains invalid numeric data."""


class InvalidCityRecordError(FileHandlingError):
    """Raised when a city record does not have valid format."""


def read_non_empty_lines(filename):
    try:
        with open(filename, "r") as file:
            lines = [line.strip() for line in file if line.strip()]
        if not lines:
            raise EmptyFileError(f"{filename} is empty.")
        return lines
    except FileNotFoundError:
        print(f"Error: {filename} file not found.")
        return []
    except PermissionError:
        print(f"Error: Permission denied for {filename}.")
        return []
    except EmptyFileError as error:
        print(f"Error: {error}")
        return []


def question_1():
    print("\n" + "=" * 70)
    print("1. OPERATIONS ON name.txt")
    print("=" * 70)

    names = read_non_empty_lines("name.txt")
    if not names:
        return

    vowel_count = 0
    vowels = "aeiou"

    for name in names:
        if name[0].lower() in vowels:
            vowel_count += 1

    longest_name = max(names, key=len)

    print("Names in file:")
    for name in names:
        print(name)

    print(f"\na. Total number of names = {len(names)}")
    print(f"b. Names starting with vowel = {vowel_count}")
    print(f"c. Longest name = {longest_name}")


def question_2():
    print("\n" + "=" * 70)
    print("2. OPERATIONS ON integers.txt")
    print("=" * 70)

    lines = read_non_empty_lines("integers.txt")
    if not lines:
        return

    numbers = []

    try:
        for line in lines:
            numbers.append(int(line))
    except ValueError as error:
        raise InvalidNumberError(f"Invalid number found in integers.txt: {error}")

    maximum_number = max(numbers)
    average_number = sum(numbers) / len(numbers)
    greater_than_100 = 0

    for number in numbers:
        if number > 100:
            greater_than_100 += 1

    print("Numbers in file:", numbers)
    print(f"a. Maximum number = {maximum_number}")
    print(f"b. Average of numbers = {average_number:.2f}")
    print(f"c. Count of numbers greater than 100 = {greater_than_100}")


def question_3():
    print("\n" + "=" * 70)
    print("3. OPERATIONS ON city.txt")
    print("=" * 70)

    lines = read_non_empty_lines("city.txt")
    if not lines:
        return

    cities = []

    try:
        for line in lines:
            parts = line.split()
            if len(parts) != 3:
                raise InvalidCityRecordError(f"Invalid city record: {line}")

            city_name = parts[0]
            population = float(parts[1])
            area = float(parts[2])
            cities.append((city_name, population, area))
    except ValueError as error:
        raise InvalidCityRecordError(f"Invalid numeric value in city.txt: {error}")

    print("a. Details of all cities:")
    for city_name, population, area in cities:
        print(f"{city_name} - Population: {population} lakhs, Area: {area} sq km")

    print("\nb. Cities with population more than 10 lakhs:")
    found = False
    for city_name, population, area in cities:
        if population > 10:
            print(city_name)
            found = True
    if not found:
        print("No city found.")

    total_area = 0
    for city_name, population, area in cities:
        total_area += area

    print(f"\nc. Sum of areas of all cities = {total_area:.2f} sq km")


def question_4():
    print("\n" + "=" * 70)
    print("4. INTEGER DIVISION WITH EXCEPTION HANDLING")
    print("=" * 70)
    print("Enter number of test cases, then enter values of a and b.")

    try:
        test_cases = int(input("Enter number of test cases: "))
    except ValueError as error:
        print(f"Error Code: {error}")
        return

    for i in range(test_cases):
        user_input = input(f"Enter a and b for test case {i + 1}: ")
        values = user_input.split()

        try:
            a = int(values[0])
            b = int(values[1])
            print(a // b)
        except ZeroDivisionError as error:
            print(f"Error Code: {error}")
        except ValueError as error:
            print(f"Error Code: {error}")
        except IndexError:
            print("Error Code: Please enter two values.")


def question_5():
    print("\n" + "=" * 70)
    print("5. MULTIPLE CUSTOM EXCEPTIONS FOR FILE HANDLING")
    print("=" * 70)

    print("Custom exceptions used in this program:")
    print("1. FileNotFoundError - when file does not exist")
    print("2. PermissionError - when file cannot be accessed")
    print("3. EmptyFileError - when file is empty")
    print("4. InvalidNumberError - when numeric file has wrong data")
    print("5. InvalidCityRecordError - when city data format is wrong")


def main():
    try:
        question_1()
        question_2()
        question_3()
        question_4()
        question_5()
    except InvalidNumberError as error:
        print(f"Error: {error}")
    except InvalidCityRecordError as error:
        print(f"Error: {error}")
    except Exception as error:
        print(f"Unexpected error: {error}")


main()
```
